```python
import pandas as pd
```

```python
countries = pd.read_csv('data/countries.csv')
cpi = pd.read_csv('data/CPI Data.csv')
tariffs = pd.read_csv('data/tarrifs-by-country.csv')
```

```python
joined = pd.merge(
    left=cpi, 
    right=tariffs,
    how='inner',
    left_on=['Country Code', 'Year'],
    right_on=['Country Code', 'Year'],
)

joined = joined.drop(
    'country',
    axis=1
)
joined
```

```python
joined[joined['CPI']<25].plot('Average Tarriff','CPI','scatter')
```
